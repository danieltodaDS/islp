### Standard Error (SE)

**Importante!**

Para ser valido essas formulas os residuos precisam: 
- ter variancia em comum (Homecedasticidade). ou seja, o ruido do modelo seja o mesmo para todas as entradas
- e serem descorrelacionados. Isso significa dizer que saber o residuo de uma entrada nao dá pista sobre os residuos de outra entrada

-----------
**leverage**

- pareceu contra-intuitivo no começo. mas os xi estarem mais espalhados em relação a media faz com que pequenas variações em y, decorrentes dos residuos (ou seja, fatores nao capturados pelo modelo) impactem menos a inclinação da curva do modelo
- com isso, torna mais "confiavel" esse estimador ---> a distancia entre o estimador e o valor real do parametro é menor

### Residual Standard Error (RSE)

- tem a ver com o fato que nao conhecemos a real variancia dos estimadores
- portanto, estimamos essa variancia usando o Residual Sum of Squares (RSS)
- RSS, por sua vez, é o somatorio dos quadrados dos residuos, sendo que os residuos são a diferenca entre o valor estimado de y e seu valor real

# Intervalos de confiança dos parametros da regressao

**intervalo de confianca**

- um intervalo de confiança de 95% significa que se reamostrarmos infinitas vezes, em 95% das vezes o intervalo deverá conter o verdadeiro valor do parametro
- um intervalo de confiança é relativo a um amostra. porem o metodo de construção garante que 95% das vezes, os intervalos gerados deverão conter o valor real do parametro

--------------
**intervalos de confiança dos estimadores da regressao**

- sao dados a partir do erro padrao calculado desses estimadores
- lembrando a formula desses estimadores é valida desde que os residuos:
    - tenham variancia constante entre todos pontos de dados
    - nao sejam correlacionados
